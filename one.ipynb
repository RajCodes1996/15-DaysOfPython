What is mutability?
- Mutable: An object can be modified after creation.
- Immutable: An object cannot be modified after creation.
Mutable	Immutable

Dictionary	Float
Set	String
Bytearray	Tuple
	Boolean

```python
# Mutable oject
numbers = [10,20,30]
numbers[0] = 100
print(numbers)

# Immutable object
name = "Python"
name = name + "Programming"
print(name)

# Notice something important: Python did not modify the original string object. It created a new string and reassigned the variable.
```

### Section 2 – Loops in Python
1. What is a loop?
A loop allows us to execute a block of code repeatedly.

Imagine you need to print the names of 100 students.

Without loops, you might write 100 print statements.

With loops, you can accomplish this using just a few lines.

Python primarily provides two types of loops:
- for loop
- while loop
2. For loop
A for loop iterates over a sequence or iterable.

Example:

```python
for i in range(5):
    print(i)
```

```python
# Understanding range()
# This is important.

# range() generates a sequence of integers.

# It has three common forms:
# Form 1: range(stop)

for i in range(5):
    print(i)

# Form 2: range(start, stop)
for i in range(1,6):
    print(i)
    
# Form 3: range(start, stop, step)
for i in range(2,11,2):
    print(i)
    
# Notice that the stop value is excluded.
```

```python
# Practical example: Multiplication table
number = 5
  
for i in range(1,11):
    print(number,"x",i, number * i)
```

```python
# 3. While loop
# A while loop executes as long as its condition remains True.

# Example:

count = 1

while count <= 5:
    print(count)
    count += 1
    
# Let's understand:
# 1. Initially, count is 1.
# 2. Python checks count <= 5.
# 3. It prints the value.
# 4. count += 1 increases the count.
# 5. The process repeats until the condition becomes False.
# (Important) Always ensure that your while loop has a way to terminate, otherwise you may create an infinite loop.
```

4. Difference between for and while

For loop: 
-Iterates over an iterable
-Useful when iterating through a sequence
-Often used with range()
-Commonly used for fixed iterations

While loop:
-Runs while a condition is True
-Useful when repetition depends on a condition
-Requires condition management
-Commonly used for condition-based iterations

Example applications:
- For loop: Process all student records.
- While loop: Keep asking for a password until correct.

### Section 3 – Loop Control Statements
Python provides three important loop control statements.

```python
# 1. break
# Immediately terminates the loop.

for i in range(1,11):
    if i == 5:
        break
    print(i)
    
# When i becomes 5, the loop stops.
```

```python
# 2. continue
# Skips the current iteration and moves to the next one.

for i in range(1,6):
    if i == 3:
        continue
    print(i)
# The number 3 is skipped.

# 3. pass
# The pass statement does nothing. It acts as a placeholder where Python expects a statement.

for i in range(5):
    if i == 2:
        pass
    print(i)
```

### Section 4 – Important Python Loop Functions
These are particularly useful when working with real datasets.

```python
# 1. enumerate()
# enumerate() provides both the index and value while iterating.

students = ["Raj","Rahul","Amit","Priya"]
for index, name in enumerate(students):
    print(index,name)
```

```python
# You can start the index from 1:
for index, name in enumerate(students,start=1):
    print(index,name)
    
# # 2. zip()
# zip() combines corresponding elements from multiple iterables.
students = ["Raj","Rahul","Amit"]
marks = [90,85,95]

for name, mark in zip(students,marks):
    print(name,mark)
    
# This is especially useful when processing related columns or collections.
# Note: By default, zip() stops when the shortest iterable is exhausted.
```

```python
# 3. Nested loops
# A nested loop means a loop inside another loop.

for i in range(1,4):
    for j in range(4,10):
        print(i,j)
        
# The inner loop completes all its iterations for every single iteration of the outer loop.
```
