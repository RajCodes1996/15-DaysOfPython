### Section 4 – Mutable vs Immutable Objects
(Important concept for trainer interviews)

This is one of the concepts you should be able to explain confidently.
What is mutability?
- Mutable: An object can be modified after creation.
- Immutable: An object cannot be modified after creation.
Mutable	Immutable
List	Integer
Dictionary	Float
Set	String
Bytearray	Tuple
	Boolean
	Bytes
	Frozenset

```python
# Mutable oject
numbers = [10,20,30]
numbers[0] = 100
print(numbers)

# Immutable object
name = "Python"
name = name + "Programming"
print(name)

# Notice something important: Python did not modify the original string object. It created a new string and reassigned the variable.
```

```python
# Understanding object identity
numbers = [10,20,30]
print(id(numbers))
numbers.append(40)
print(id(numbers))

# The identity remains the same because the list itself was modified.
```

```python
# Now compare:
name = "Python"
print(id(name))
name = name + "Programming"
print(id(name))

# The new string will have a different identity.

# Note: id() returns an object's identity during its lifetime. Don't rely on particular numeric ID values, This is callled Immutability!
```

### Section 5 – Python Operators
Python operators are symbols or keywords used to perform operations.

```python
# 1. Arithmetic operators
a = 10
b = 3
print(a + b) # Addition
print(a - b) # Subtraction
print(a * b) # Multiplication
print(a / b) # Division
print(a //b) # Floor Division
print(a % b) # Modulus
print(a ** b) # Exponentiation
 
#  2. Comparison operators
a = 10
b = 20
print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
# These return Boolean values.

# 3. Logical operators
age = 25
has_id = True
print(age >= 18 and has_id)
print(age >= 18 or has_id)
print(not has_id)

# Operators:
# - and: Both conditions must be true.
# - or: At least one condition must be true.
# - not: Reverses the Boolean result.

# 4. Assignment operators
x = 10
x += 5
x -= 2
x *= 2
print(x)

# 5. Membership operators
languages = ["python", "Java", "C++"]
print("Python" in languages)
print("PHP"  not in languages)

# 6. Identity operators
a = [1,2,3]
b = a
c = [1,2,3]
print(a == c)
print(a is c)
print(a is b)

# 7. Operator precedence
result = 10 + 5 * 2
print(result)
```

### Section 6 – Practical Implementation
Now let's combine today's concepts into one working program.
Project: Student Information & Result Analyzer
Objective: Build a Python program that accepts student information, calculates marks, determines the result and displays a formatted report.

```python
# DAY 1 - Student Information & Result Analyzer

# 1. Taking student information
name = input("Enter student name: ")
age = int(input("Enter student age: "))

# 2. Taking subject marks
python_marks = float(input("Enter Python marks: "))
ml_marks = float(input("Enter ML marks: "))
cs_marks = float(input("Enter CS marks: "))

# 3. Calculating total and percentage
total = python_marks + ml_marks + cs_marks
percentage = total / 3

# 4. Checking pass/fail
is_passed = (
    python_marks >= 40
    and ml_marks >= 40
    and cs_marks >= 40
)

# 5. Assigning grade: 
if percentage >= 90:
    grade = "A+"
elif percentage >= 80:
    grade = "A"
elif percentage >= 70:
    grade = "B"
elif percentage >= 60:
    grade = "C"
    
else : 
    grade = "Needs Improvement"

# 6. Displaying report
print("\n========== STUDENT REPORT ==========")

print(f"Student Name: {name}")
print(f"Age: {age}")
print(f"Total Marks: {total}/300")
print(f"Percentage: {percentage:.2f}%")
print(f"Grade: {grade}")

print("Result:", "PASS" if is_passed else "FAIL")

print("====================================")
```
