Section 1 – Understanding Python Data Structures
Before studying each data structure, let's understand why we need them.

Imagine you are developing a college management application.

You need to store:
- Student names.
- Student marks.
- Student contact details.
- Unique student IDs.
- Subject information.
Creating separate variables for everything would be inefficient.

Instead, Python provides collections that allow us to store multiple values in a single variable.

There are four major collection types:

1. List

Ordered, mutable, allows duplicates.

Example: [10, 20, 30]
2. Tuple

Ordered, immutable, allows duplicates.

Example: (10, 20, 30)
3. Set

Unordered, mutable, stores unique elements.

Example: {10, 20, 30}
4. Dictionary

Stores key-value pairs, insertion-ordered, mutable.

Example: {"name": "Raj", "age": 25}
